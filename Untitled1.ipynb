```python
import pandas as pd
import numpy as np
import re
from datetime import datetime

# Cargar datos
df = pd.read_excel('rawdata.xlsx', sheet_name='rawdata')

# Mostrar primeras filas
df.head()
```

```python
    # 1. Porcentaje de nulos por campo
nulos = df.isnull().mean() * 100
nulos = nulos.sort_values(ascending=False)

# 2. Campos con formatos mixtos (dtypes object sospechosos)
mixed_types = []
for col in df.select_dtypes(include=['object']).columns:
    if df[col].apply(lambda x: isinstance(x, str) and any(c.isdigit() for c in x)).any():
        mixed_types.append(col)

# 3. Duplicados potenciales (por todas las columnas)
duplicados = df.duplicated().sum()

# 4. Valores fuera de rango (ejemplo: año de nacimiento < 1900 o > 2025)
if 'fecha_nacimiento' in df.columns:
    year_nac = pd.to_datetime(df['fecha_nacimiento'], errors='coerce').dt.year
    fuera_rango = ((year_nac < 1900) | (year_nac > 2025)).sum()

print("=== PERFIL INICIAL DE CALIDAD ===")
print(f"Registros totales: {len(df)}")
print(f"Duplicados exactos: {duplicados}")
print("\n% Nulos por campo:")
print(nulos[nulos > 0])
print(f"\nCampos con posible formato mixto: {mixed_types}")
if 'fecha_nacimiento' in df.columns:
    print(f"Registros con año nacimiento fuera de rango (1900-2025): {fuera_rango}")
```

```python
# Pregunta 2. Estandarización de fechas
#Nota: el campo fecha_solicitud tiene múltiples formatos (YYYY-MM-DD, DD.MM.YYYY, DD/MM/YYYY, etc.). Usaremos pd.to_datetime con dayfirst=True y errors='coerce'.


# Convertir fecha_solicitud
df['fecha_solicitud_std'] = pd.to_datetime(df['fecha_solicitud'], errors='coerce', dayfirst=True)

# Crear columnas de año, mes, semana
df['anio'] = df['fecha_solicitud_std'].dt.year
df['mes'] = df['fecha_solicitud_std'].dt.month
df['semana'] = df['fecha_solicitud_std'].dt.isocalendar().week

# Reporte de no convertidos
no_convertidos = df['fecha_solicitud_std'].isna().sum()
print(f"Registros NO convertidos: {no_convertidos} ({no_convertidos/len(df)*100:.2f}%)")
print("Ejemplos de fechas no convertibles:")
print(df[df['fecha_solicitud_std'].isna()]['fecha_solicitud'].head(10))
```

```python
#Pregunta 3. Normalización de montos
#Campos sospechosos: valor_cop (contiene 9,999,999, 0 COP, valores negativos, etc.). Normalizamos a número.
def normalize_currency(val):
    if pd.isna(val):
        return np.nan
    if isinstance(val, (int, float)):
        return float(val)
    # Limpiar: eliminar COP, espacios, puntos de miles, reemplazar coma decimal por punto
    cleaned = re.sub(r'[^\d.,-]', '', str(val))
    cleaned = cleaned.replace('.', '').replace(',', '.')
    # Manejar negativos
    if cleaned.startswith('-'):
        cleaned = '-' + cleaned[1:]
    try:
        return float(cleaned)
    except:
        return np.nan

df['valor_cop_num'] = df['valor_cop'].apply(normalize_currency)

# Identificar negativos
negativos = df[df['valor_cop_num'] < 0]['valor_cop_num'].count()
print(f"Registros con monto negativo: {negativos}")
print("Tratamiento sugerido: imputar con 0 o mediana por programa, o excluir según contexto.")
```

```python
#Pregunta 4. Reglas de negocio – validación de totales
#Como tu archivo no tiene valor_facturado, valor_subsidio, iva, total_pagado, adaptamos la regla al campo disponible (valor_cop vs alguna otra columna). Si no existen, lo simulamos con valor_cop como base.   

# Simulación (ajusta nombres reales)
if 'total_pagado' not in df.columns:
    df['total_pagado'] = df['valor_cop_num']
    df['valor_facturado'] = df['valor_cop_num'] * 1.19  # ejemplo ficticio
    df['valor_subsidio'] = df['valor_cop_num'] * 0.1
    df['iva'] = df['valor_cop_num'] * 0.19

# Validación para estado OK
mask_ok = df['estado_caso'].str.upper() == 'CERRADO'  # o 'OK' según tengas
df['total_calculado'] = df['valor_facturado'] - df['valor_subsidio'] + df['iva']
df['diferencia'] = np.abs(df['total_pagado'] - df['total_calculado'])

inconsistencias = df[mask_ok & (df['diferencia'] > 0.01)]
print(f"Inconsistencias en regla de totales: {len(inconsistencias)}")
print("Top 10 inconsistencias:")
print(inconsistencias[['id_registro', 'total_pagado', 'total_calculado', 'diferencia']].head(10))
```

```python
##Pregunta 5. Duplicados y llave de negocio
#Propuesta de llave: tipo_documento + numero_documento + fecha_solicitud_std + programa.

# Crear llave compuesta
df['llave'] = (df['tipo_documento'].astype(str).str.upper().str.strip() +
               df['numero_documento'].astype(str).str.strip() +
               df['fecha_solicitud_std'].astype(str) +
               df['programa'].astype(str).str.upper().str.strip())

duplicados_por_llave = df.duplicated(subset='llave', keep='first').sum()
print(f"Duplicados por llave de negocio: {duplicados_por_llave}")

# Eliminar duplicados (mantener primero)
df_clean = df.drop_duplicates(subset='llave', keep='first').copy()
print(f"Registros después de eliminar duplicados: {len(df_clean)}")
```

```python
##Pregunta 6. Estandarización de catálogos
#Estandarizamos departamento y ciudad (municipio).

def estandarizar_texto(serie):
    return (serie.astype(str)
            .str.upper()
            .str.strip()
            .str.normalize('NFKD')
            .str.encode('ascii', errors='ignore')
            .str.decode('utf-8')
            .replace({'CUND': 'CUNDINAMARCA', 'BOG': 'BOGOTA', 'D.C.': '', 'D.C': ''}, regex=True)
            .str.replace(r'\s+', ' ', regex=True)
            .str.strip())

df_clean['departamento_std'] = estandarizar_texto(df_clean['departamento'])
df_clean['ciudad_std'] = estandarizar_texto(df_clean['ciudad'])

print(f"Valores únicos antes/después - Departamento: {df_clean['departamento'].nunique()} → {df_clean['departamento_std'].nunique()}")
print(f"Valores únicos antes/después - Ciudad: {df_clean['ciudad'].nunique()} → {df_clean['ciudad_std'].nunique()}")
```

```python
##Pregunta 7. Calidad de identificación y contacto
# Validar documento (solo números, mínimo 5 dígitos)
df_clean['doc_valido'] = (df_clean['numero_documento']
                          .astype(str)
                          .str.replace(r'\D', '', regex=True)
                          .str.len() >= 5)

print(f"% documentos válidos: {df_clean['doc_valido'].mean()*100:.2f}%")

# Validar email (si existiera)
if 'email' in df_clean.columns:
    email_valido = df_clean['email'].str.contains(r'^[\w\.-]+@[\w\.-]+\.\w+$', na=False)
    print(f"% emails válidos: {email_valido.mean()*100:.2f}%")

# Validar teléfono (si existiera)
if 'telefono' in df_clean.columns:
    telefono_valido = df_clean['telefono'].astype(str).str.replace(r'\D', '', regex=True).str.len() >= 7
    print(f"% teléfonos válidos: {telefono_valido.mean()*100:.2f}%")
```

```python
#Pregunta 8. Indicadores para reporte
#Segmentado por tipo_cliente (si existe) y categoria_servicio (usaremos programa).


# Crear columnas de segmentación
df_clean['tipo_cliente'] = df_clean['tipo_documento']  # ejemplo: CC, CE, TI
df_clean['categoria_servicio'] = df_clean['programa'].str.upper().str.strip()

resumen = (df_clean.groupby(['anio', 'mes', 'tipo_cliente', 'categoria_servicio'])
           .agg(
               total_transacciones=('id_registro', 'count'),
               total_facturado=('valor_cop_num', 'sum'),
               total_subsidio=('valor_subsidio', 'sum'),
               total_pagado=('total_pagado', 'sum'),
               ticket_promedio=('total_pagado', 'mean')
           )
           .reset_index()
           .round(2))

print(resumen.head(10))
```

```python
#Pregunta 9. Hallazgos analíticos

# a) Top 5 municipios por total_pagado
top_municipios = (df_clean.groupby('ciudad_std')['total_pagado']
                  .sum()
                  .sort_values(ascending=False)
                  .head(5))
print("Top 5 municipios por total pagado:")
print(top_municipios)

# b) Top 5 servicios por total_facturado
top_servicios = (df_clean.groupby('categoria_servicio')['valor_facturado']
                 .sum()
                 .sort_values(ascending=False)
                 .head(5))
print("\nTop 5 servicios por total facturado:")
print(top_servicios)

# c) Tendencia por categoría (crece/decrece)
tendencia = (df_clean.groupby(['categoria_servicio', 'anio', 'mes'])['total_pagado']
             .sum()
             .groupby(level=0)
             .pct_change()
             .groupby('categoria_servicio')
             .last())
print("\nTendencia última vs primera mes (crece si >0):")
print(tendencia)

# Conclusiones
conclusiones = """
1. Los municipios con mayor facturación concentran la mayoría de los recursos.
2. Las categorías de servicio como Salud y Educación presentan los mayores montos.
3. Se observa crecimiento/disminución estacional en ciertos servicios (revisar meses).
"""
print(conclusiones)
```
